```python
cars_per_user = (
    dem.groupby('client_id')['auto_document_id']
       .nunique()
       .reset_index(name='n_cars')
)

dist = (cars_per_user['n_cars']
        .value_counts()
        .sort_index()
        .rename_axis('n_cars')
        .reset_index(name='n_users'))

# ---------- график ----------
palette = [PINK, ORANGE, PURPLE]
colors = [palette[i % 3] for i in range(len(dist))]

fig, ax = plt.subplots(figsize=(9, 5))

bars = ax.bar(dist['n_cars'], dist['n_users'],
              color=colors, edgecolor=WHITE, linewidth=1.2)

for bar, val in zip(bars, dist['n_users']):
    ax.text(bar.get_x() + bar.get_width() / 2,
            bar.get_height(),
            f'{val:,}'.replace(',', ' '),
            ha='center', va='bottom', fontsize=10, color=INK)

ax.set_xlabel('Количество машин у пользователя')
ax.set_ylabel('Количество пользователей')
ax.set_title('Распределение числа машин на пользователя')
ax.set_xticks(dist['n_cars'])

plt.tight_layout()
plt.show()
```

```python
import matplotlib.dates as mdates

# ---------- 1. Готовим данные ----------
df = fuel.copy()
df['order_datetime'] = pd.to_datetime(df['order_datetime'])
df['date'] = df['order_datetime'].dt.normalize()

daily = (df.groupby('date')
           .size()
           .rename('n')
           .reset_index()
           .sort_values('date'))

# ---------- 2. Определяем неполные месяцы ----------
# Неполные = первый и последний месяц в данных
first_month = daily['date'].min().to_period('M')
last_month  = daily['date'].max().to_period('M')

daily['month'] = daily['date'].dt.to_period('M')
daily['is_incomplete'] = daily['month'].isin([first_month, last_month])
daily['color'] = np.where(daily['is_incomplete'], GREY, PURPLE)

# ---------- 3. График ----------
fig, ax = plt.subplots(figsize=(14, 5))

ax.bar(daily['date'], daily['n'],
       color=daily['color'], width=1.0, linewidth=0)

# ---------- 4. Пунктирные границы месяцев ----------
month_starts = pd.date_range(daily['date'].min().to_period('M').start_time,
                             daily['date'].max().to_period('M').end_time,
                             freq='MS')
month_ends = month_starts + pd.offsets.MonthEnd(0)

for d in list(month_starts) + list(month_ends):
    ax.axvline(d, color=MUTE, linestyle=':', linewidth=0.8, alpha=0.7)

# ---------- 5. Метки месяцев на оси X ----------
ax.xaxis.set_major_locator(mdates.MonthLocator())
ax.xaxis.set_major_formatter(mdates.DateFormatter('%b\n%Y'))

# ---------- 6. Легенда «Неполные месяцы» ----------
from matplotlib.patches import Patch
legend_handles = [Patch(facecolor=GREY, label='Неполные месяцы (март, сентябрь)')]
ax.legend(handles=legend_handles, loc='upper left', frameon=False)

# ---------- 7. Подписи ----------
ax.set_xlabel('Дата')
ax.set_ylabel('Число записей в день')
ax.set_title('Распределение данных по дням: март и сентябрь — неполные')

plt.tight_layout()
plt.show()
```

```python
# ---------- 1. Готовим данные ----------
fines['bill_offence_date'] = pd.to_datetime(fines['bill_offence_date'])
fines['date'] = fines['bill_offence_date'].dt.normalize()
fines['dow'] = fines['bill_offence_date'].dt.dayofweek      # 0=Пн, 6=Вс
fines['dow_name'] = fines['bill_offence_date'].dt.day_name()
fines['week'] = fines['bill_offence_date'].dt.to_period('W').dt.start_time

# проверяем первую дату
print("Первая запись:", fines['bill_offence_date'].min())
print("День недели первой записи:", fines['bill_offence_date'].min().day_name())

# ---------- 2. Распределение по дням недели ----------
order = ['Monday', 'Tuesday', 'Wednesday', 'Thursday', 'Friday', 'Saturday', 'Sunday']

by_dow = (fines['dow_name']
          .value_counts()
          .reindex(order))

print("\nПо дням недели:")
print(by_dow)

fig, ax = plt.subplots(figsize=(10, 5))

colors_dow = [PINK if d in ('Saturday', 'Sunday') else PURPLE for d in order]
ax.bar(range(7), by_dow.values, color=colors_dow, edgecolor=WHITE, linewidth=1.2)

ax.set_xticks(range(7))
ax.set_xticklabels(['Пн', 'Вт', 'Ср', 'Чт', 'Пт', 'Сб', 'Вс'])
ax.set_ylabel('Число штрафов')
ax.set_title('Распределение штрафов по дням недели')

plt.tight_layout()
plt.show()

pivot = (fines.groupby(['week', 'dow_name'])
              .size()
              .unstack(fill_value=0)
              .reindex(columns=order, fill_value=0))

fig, ax = plt.subplots(figsize=(14, 6))

palette_dow = [PURPLE, ORANGE, GREEN, PINK, '#00A88F', '#FD8616', '#7061F0']
for i, d in enumerate(order):
    ax.plot(pivot.index, pivot[d], color=palette_dow[i], label=d, linewidth=2.0)

ax.set_xlabel('Неделя')
ax.set_ylabel('Число штрафов за день')
ax.set_title('Штрафы по дням недели: динамика по неделям')
ax.legend(loc='upper right', frameon=False, ncol=7, fontsize=10)

plt.tight_layout()
plt.show()

share = pivot.div(pivot.sum(axis=1), axis=0)
print(share.describe().T[['mean', 'std', 'min', 'max']])
```

```python
# ---------- 1. Подготовка ----------
fines['bill_offence_date'] = pd.to_datetime(fines['bill_offence_date'])
fines['date'] = fines['bill_offence_date'].dt.normalize()
fines['dow_name'] = fines['bill_offence_date'].dt.day_name()

order = ['Monday', 'Tuesday', 'Wednesday', 'Thursday', 'Friday', 'Saturday', 'Sunday']
labels = ['Пн', 'Вт', 'Ср', 'Чт', 'Пт', 'Сб', 'Вс']

# ---------- 2. Среднее на один день недели ----------
# сколько каждого дня недели реально было в периоде
cal = pd.date_range(fines['date'].min(), fines['date'].max(), freq='D')
n_days_of_dow = (pd.Series(cal)
                 .dt.day_name()
                 .value_counts()
                 .reindex(order))

by_dow_sum  = fines['dow_name'].value_counts().reindex(order)
by_dow_mean = (by_dow_sum / n_days_of_dow).round(2)

# ---------- 3. Медиана дневных значений ----------
daily_counts = fines.groupby('date').size().rename('n').reset_index()
daily_counts['dow_name'] = daily_counts['date'].dt.day_name()

by_dow_median = (daily_counts
                 .groupby('dow_name')['n']
                 .median()
                 .reindex(order))

# ---------- 4. Два графика рядом ----------
fig, axes = plt.subplots(1, 2, figsize=(16, 5), sharey=False)

colors_dow = [PINK if d in ('Saturday', 'Sunday') else PURPLE for d in order]

# --- слева: среднее ---
axes[0].bar(range(7), by_dow_mean.values,
            color=colors_dow, edgecolor=WHITE, linewidth=1.2)
axes[0].set_xticks(range(7))
axes[0].set_xticklabels(labels)
axes[0].set_ylabel('Среднее число штрафов в день')
axes[0].set_title('Среднее по дням недели')

# --- справа: медиана ---
axes[1].bar(range(7), by_dow_median.values,
            color=colors_dow, edgecolor=WHITE, linewidth=1.2)
axes[1].set_xticks(range(7))
axes[1].set_xticklabels(labels)
axes[1].set_ylabel('Медианное число штрафов в день')
axes[1].set_title('Медиана по дням недели')

plt.tight_layout()
plt.show()

# ---------- 5. Таблица для сверки ----------
comparison = pd.DataFrame({
    'mean':   by_dow_mean.values,
    'median': by_dow_median.values,
}, index=labels)
comparison['diff'] = (comparison['mean'] - comparison['median']).round(1)
print(comparison)
```

```python
import numpy as np
import matplotlib.pyplot as plt

order  = ['Monday', 'Tuesday', 'Wednesday', 'Thursday', 'Friday', 'Saturday', 'Sunday']
labels = ['Пн', 'Вт', 'Ср', 'Чт', 'Пт', 'Сб', 'Вс']

median_vals = by_dow_median.values

# baseline — среднее по семи медианам
baseline = median_vals.mean()

# отклонение каждого дня от baseline, в % от baseline
dev_pct = (median_vals - baseline) / baseline * 100

colors_dow = [PINK if d in ('Saturday', 'Sunday') else PURPLE for d in order]

# ---------- График: столбики + линия baseline + подписи отклонений ----------
fig, ax = plt.subplots(figsize=(11, 6))

bars = ax.bar(range(7), median_vals, color=colors_dow,
              edgecolor=WHITE, linewidth=1.2)

# линия среднего
ax.axhline(baseline, color=INK, linestyle='--', linewidth=1.3,
           label=f'Среднее по неделе = {baseline:.1f}')

# подписи: и абсолютное значение, и отклонение в %
for i, (v, dev) in enumerate(zip(median_vals, dev_pct)):
    ax.annotate(f'{v:.0f}',
                (i, v), textcoords='offset points', xytext=(0, 22),
                ha='center', fontsize=10, color=INK)
    ax.annotate(f'{dev:+.1f}%',
                (i, v), textcoords='offset points', xytext=(0, 8),
                ha='center', fontsize=9,
                color=(PINK if dev > 0 else '#00A88F'))

ax.set_xticks(range(7))
ax.set_xticklabels(labels)
ax.set_ylabel('Медианное число штрафов в день')
ax.set_title('Медиана по дням недели: отклонение от среднего')
ax.legend(frameon=False, loc='upper right')

plt.tight_layout()
plt.show()
```

```python
fig, ax = plt.subplots(figsize=(11, 5))

# столбики от нуля, цвет = знак
bar_colors = [PINK if d > 0 else '#00A88F' for d in dev_pct]
ax.bar(range(7), dev_pct, color=bar_colors, edgecolor=WHITE, linewidth=1.2)

ax.axhline(0, color=INK, linewidth=1.0)
ax.set_xticks(range(7))
ax.set_xticklabels(labels)
ax.set_ylabel('Отклонение от среднего по неделе, %')
ax.set_title('Отклонение медианы каждого дня от среднего по неделе')

for i, d in enumerate(dev_pct):
    ax.annotate(f'{d:+.1f}%', (i, d),
                textcoords='offset points',
                xytext=(0, 6 if d > 0 else -14),
                ha='center', fontsize=10, color=INK)

plt.tight_layout()
plt.show()
```
